# K-NeighborsClassifier from scratch

## Define our own class for KNN

```python
import numpy as np
from collections import Counter

class KNN:
    def __init__(self, n_neighbors=5):
        self.k = n_neighbors
        self.x_train = None
        self.y_train = None

    def fit(self, x_train, y_train):
        self.x_train = x_train
        self.y_train = y_train

    def predict(self, x_test):
        y_pred = []
        for i in x_test:
            distances = []
            for j in self.x_train:
                distances.append(self.calculate_distance(i, j))
            sorted_distances = sorted(list(enumerate(distances)), key = lambda x:x[1])
            n_neighbor = sorted_distances[0 : self.k]
            labels = self.majority_count(n_neighbor)
            y_pred.append(labels)
        return np.array(y_pred)

    def calculate_distance(self, pointA, pointB):
        return np.linalg.norm(pointA - pointB)

    def majority_count(self, neighbors):
        votes = []
        for i in neighbors:
            votes.append(y_train[i[0]])

        votes = Counter(votes)
        return votes.most_common()[0][0]
```

### Splitting and Scaling data

```python
x = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=19)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

### Using own KNN class (model)

```python
apnaKNN = KNN(n_neighbors=3)
```

```python
apnaKNN.fit(x_train_scaled, y_train)
```

```python
y_pred = apnaKNN.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test, y_pred)
```

### Comparing with Scikit-Learn KNN class

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn_skl = KNeighborsClassifier(n_neighbors=3)
```

```python
knn_skl.fit(x_train_scaled, y_train)
y_pred1 = knn_skl.predict(x_test_scaled)
```

```python
accuracy_score(y_test, y_pred1)
```
