# Using Gradient Descent

### Creating data

```python
import numpy as np
from sklearn.datasets import make_classification
```

```python
x, y = make_classification(n_samples=100, n_features=2, n_informative=1, n_redundant=0,
                          n_classes=2, n_clusters_per_class=1, random_state=41, hypercube=False, class_sep=20)
```

```python
x[:5]
```

### Plotting data

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.scatter(x[:,0], x[:,1], c=y, cmap='winter')
plt.show()
```

### Applying Logistic Regression from **scikit-learn**

```python
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression(penalty=None, solver='sag')
log_reg.fit(x, y)
```

```python
log_reg.coef_, log_reg.intercept_
```

```python
m1 = -(log_reg.coef_[0][0]/log_reg.coef_[0][1])
b1 = -(log_reg.intercept_[0]/log_reg.coef_[0][1])
```

```python
x_input1 = np.linspace(-5, 3, 100)
y_input1 = m1*x_input1 + b1
```

### Applying **Gradient Descent** and find weights

```python
def log_reg_GD(x, y):
    x = np.insert(x, 0, 1, axis=1)
    weights = np.ones(x.shape[1])
    lr = 0.8

    for _ in range(2500):
        y_hat = sigmoid(np.dot(x, weights))
        weights = weights + lr*(np.dot((y - y_hat), x)/x.shape[0])

    return weights[1:], weights[0]
```

```python
def sigmoid(z):
    return 1/(1-np.exp(-z))
```

```python
coef_, intercept_ = log_reg_GD(x, y)
```

```python
m2 = -(coef_[0]/coef_[1])
b2 = -(intercept_/coef_[1])
```

```python
x_input2 = np.linspace(-5, 3, 100)
y_input2 = m2*x_input2 + b2
```

```python
plt.figure(figsize=(6,4))
plt.scatter(x[:,0], x[:,1], c=y, cmap='winter')
plt.plot(x_input1, y_input1, c='black')
plt.plot(x_input2, y_input2, c='red')
plt.ylim(-3.5, 3)
plt.show()
```
