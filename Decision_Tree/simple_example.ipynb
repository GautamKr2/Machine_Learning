# Simple example of DT Classifier

### Loading dataset

```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('E:/Dataset/Social_Network_Ads.csv')
```

```python
data.head(5)
```

```python
data.isnull().sum()
```

### splitting and encoding data

```python
from sklearn.model_selection import train_test_split
```

```python
x = data.iloc[:, 1:-1]
y = data.iloc[:, -1]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=20)
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
encoder = OrdinalEncoder(categories=[['Male', 'Female']])
```

```python
x_train['Gender'] = encoder.fit_transform(x_train[['Gender']])
x_test['Gender'] = encoder.transform(x_test[['Gender']])
```

```python
encoder.categories_
```

```python
x_train.head()
```

### Model training and finding accuracy

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

#### Training model on different criterion

```python
dt_clasifier = DecisionTreeClassifier(criterion='gini')
dt_clasifier.fit(x_train, y_train)
y_pred = dt_clasifier.predict(x_test)
accuracy_score(y_test, y_pred)
```

```python
dt_clasifier1 = DecisionTreeClassifier(criterion='entropy')
dt_clasifier1.fit(x_train, y_train)
y_pred = dt_clasifier1.predict(x_test)
accuracy_score(y_test, y_pred)
```

```python
dt_clasifier = DecisionTreeClassifier(criterion='log_loss')
dt_clasifier.fit(x_train, y_train)
y_pred = dt_clasifier.predict(x_test)
accuracy_score(y_test, y_pred)
```

#### Adding max_depth hyperparameter

```python
dt_clasifier = DecisionTreeClassifier(criterion='gini', max_depth=2)
dt_clasifier.fit(x_train, y_train)
y_pred = dt_clasifier.predict(x_test)
accuracy_score(y_test, y_pred)
```

```python
dt_clasifier = DecisionTreeClassifier(criterion='entropy', max_depth=2)
dt_clasifier.fit(x_train, y_train)
y_pred = dt_clasifier.predict(x_test)
accuracy_score(y_test, y_pred)
```

```python
dt_clasifier = DecisionTreeClassifier(criterion='log_loss', max_depth=2)
dt_clasifier.fit(x_train, y_train)
y_pred = dt_clasifier.predict(x_test)
accuracy_score(y_test, y_pred)
```

## - Final conclusion is that on this dataset -
#### i) it does not depends on any criterion, it depends only on max_depth
#### ii) max_depth=2 is the best decision making boundary.
