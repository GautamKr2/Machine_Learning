### 1. Working on **Classification problem**

```python
iris_data = load_iris()
```

```python
x = iris_data.data
y = iris_data.target
```

```python
from sklearn.tree import DecisionTreeClassifier
dt_clf = DecisionTreeClassifier(max_depth=None)
```

```python
dt_clf.fit(x, y)
```

```python
dt_clf.get_depth()
```

```python
from sklearn.tree import plot_tree
plot_tree(dt_clf)
plt.show()
```

## Visualization using **dtreeviz**

```python
iris_data.feature_names
```

```python
iris_data.target_names
```

```python
viz = model(dt_clf, x, y, feature_names=iris_data.feature_names, 
              class_names=iris_data.target_names)

viz.view(scale=2)
```

### 3. **Horizontal** Decision Tree

```python
viz = model(dt_clf, x, y, feature_names=iris_data.feature_names, 
              class_names=iris_data.target_names)

viz.view(scale=1, orientation='LR')  # 'TD for top to bottom'
```

### 4. Show path for new data

```python
new_pt = iris_data.data[np.random.randint(0, len(iris_data.data)), :]

viz = model(dt_clf, x, y, feature_names=iris_data.feature_names, 
              class_names=iris_data.target_names)

viz.view(x=new_pt, scale=1)
```

```python
new_pt
```

### 5. Show node number

```python
viz = model(dt_clf, x, y, feature_names=iris_data.feature_names, 
              class_names=iris_data.target_names)

viz.view(scale=1, orientation='LR', show_node_labels=True)
```

### 6. Without any graph

```python
viz = model(dt_clf, x, y, feature_names=iris_data.feature_names, 
              class_names=iris_data.target_names)

viz.view(scale=1, orientation='LR', show_node_labels=True, fancy=False)
```

### 7. Show just path

```python
new_pt1 = iris_data.data[np.random.randint(0, len(iris_data.data)), :]

viz = model(dt_clf, x, y, feature_names=iris_data.feature_names, 
              class_names=iris_data.target_names)

viz.view(x=new_pt1, scale=1, show_just_path=True)
```

### 8. Prediction path in **plain english**

```python
from dtreeviz.interpretation import explain_prediction_plain_english
```

```python
viz_model = model(dt_clf, x, y, feature_names=iris_data.feature_names,
                    target_name="iris",
                    class_names=iris_data.target_names
                )

print(viz_model.explain_prediction_path(new_pt))
```

### 9. Feature importance

```python
viz = model(dt_clf, x, y, feature_names=iris_data.feature_names,
                class_names=iris_data.target_names
            )

print(viz.explain_prediction_path(new_pt1))
```

```python
from dtreeviz.interpretation import explain_prediction_sklearn_default

viz_model = model(dt_clf, x, y, feature_names=iris_data.feature_names,
                    class_names=iris_data.target_names
                 )

explain_prediction_sklearn_default(
    viz_model.shadow_tree,
    new_pt1
)
```
