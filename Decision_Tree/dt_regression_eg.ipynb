# Decision Tree **Regression** example

### Loading data

```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_california_housing
```

```python
data = fetch_california_housing()
```

```python
data.feature_names
```

```python
data.target_names
```

```python
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
df.head()
```

### Splitting data

```python
x = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

```python
x.sample(2)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=20)
```

### Training model

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
dt_regressor = DecisionTreeRegressor()
```

```python
dt_regressor.fit(x_train, y_train)
```

```python
y_pred = dt_regressor.predict(x_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test, y_pred)
```

### Visualizing tree

```python
dt_regressor.get_depth()
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree
```

```python
plt.figure(figsize=(8,5))
plot_tree(dt_regressor, feature_names=x.columns, max_depth=2, filled=True)
plt.show()
```

## Hyperparameter Tuning

```python
dt_reg = DecisionTreeRegressor(criterion='poisson', max_depth=9)  # criterion = "squared_error", "absolute_error"
dt_reg.fit(x_train, y_train)
r2_score(y_test, dt_reg.predict(x_test))
```

### using Grid Search CV

```python
grid_parameter = {
    'criterion': ['squared_error', 'absolute_error', 'poisson'],
    'max_depth': [3, 5, 7, 8, 9, 10, None],
    'max_features': [0.25, 0.5, 1.0],
    'min_samples_split': [0.25, 0.5, 1.0]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
dt_reg_grid = GridSearchCV(DecisionTreeRegressor(), param_grid=grid_parameter)
```

```python
dt_reg_grid.fit(x_train, y_train)
```

```python
dt_reg_grid.best_params_
```

```python
dt_reg_grid.best_score_
```

### Features importance

```python
for importance, name in sorted(zip(dt_reg.feature_importances_, x_train.columns), reverse=True):
    print(name, importance)
```
