# Elastic Net Regularization

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.metrics import r2_score
```

```python
x, y = load_diabetes(return_X_y=True)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=15)
```

### Linear Regression

```python
lin_reg = LinearRegression()
lin_reg.fit(x_train, y_train)
r2_score(lin_reg.predict(x_test), y_test)
```

### Ridge Regularization

```python
ridge = Ridge(alpha=0.00001)
ridge.fit(x_train, y_train)
r2_score(ridge.predict(x_test), y_test)
```

### Lasso Regularization

```python
lasso = Lasso(alpha=0.001)
lasso.fit(x_train, y_train)
r2_score(lasso.predict(x_test), y_test)
```

### Lasso Regularization

```python
elastic = ElasticNet(alpha=0.00001, l1_ratio=0.95)
elastic.fit(x_train, y_train)
r2_score(elastic.predict(x_test), y_test)
```
