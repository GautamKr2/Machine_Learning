#### For Outlook feature

```python
pd.crosstab(data['outlook'], data['play'])
```

```python
P_Overcast_Yes = 4/9
P_Rain_Yes = 3/9
P_Sunny_Yes = 2/9

P_Overcast_No = 0
P_Rain_No = 2/5
P_Sunny_No = 3/5
```

#### For Temperature

```python
pd.crosstab(data['temp'], data['play'])
```

```python
P_Cool_Yes = 3/9
P_Hot_Yes = 2/9
P_Mild_Yes = 4/9

P_Cool_No = 1/5
P_Hot_No = 2/5
P_Mild_No = 2/5
```

#### For Wind

```python
pd.crosstab(data['wind'], data['play'])
```

```python
P_Strong_Yes = 3/9
P_Weak_Yes = 6/9

P_Strong_No = 3/5
P_Weak_No = 2/5
```

#### For Humidity

```python
pd.crosstab(data['humidity'], data['play'])
```

```python
P_High_Yes = 3/9
P_Normal_Yes = 6/9

P_High_No = 4/5
P_Normal_No = 1/5
```

### Solution part

```python
# Outlook=Sunny, Temp=Hot, Humidity=High, Wind=Weak

Prob_Yes = (P_Sunny_Yes * P_Hot_Yes * P_High_Yes * P_Weak_Yes) * P_Yes
print(f"Probability of yes =", Prob_Yes)

Prob_No = (P_Sunny_No * P_Hot_No * P_High_No * P_Weak_No) * P_No
print(f"Probability of No =", Prob_No)
```

#### That means for problem-1 player will not play

```python
# Outlook=Overcast, Temp=Cool, Humidity=Normal, Wind=Weak

Prob_Yes2 = (P_Overcast_Yes * P_Cool_Yes * P_Normal_Yes * P_Weak_Yes) * P_Yes
print(f"Probability of yes =", Prob_Yes2)

Prob_No2 = (P_Overcast_No * P_Cool_No * P_Normal_No * P_Weak_No) * P_No
print(f"Probability of No =", Prob_No2)
```

#### That means for problem-2 players will play
