#### Encode the output column as a number

```python
encoder = LabelEncoder()
```

#### Remove HTML tags from reviews

```python
import re  # Regular Expression
clean = re.compile("<.*?>")
re.sub(clean, ' ', df.iloc[2].review)
```

```python
# Function to clean HTML Tags

def clean_html(text):
    clean = re.compile("<.*?>")
    return re.sub(clean, ' ', text)
```

```python
df['cleaned_review'] = df['review'].apply(clean_html)
```

```python
df.sample(2)
```

#### Converting all the letters into lower case

```python
# Function to convert to lower case

def convert_lower(text):
    return text.lower()
```

```python
df['cleaned_review'] = df['cleaned_review'].apply(convert_lower)
```

#### Removing special characers (₹, #, &) from reviews

```python
# Function to remove special character

def remove_special(text):
    x = ''
    for ch in text:
        if ch.isalnum():
            x = x+ch
        else:
            x = x+' '
            
    return x
```

```python
df['cleaned_review'] = df['cleaned_review'].apply(remove_special)
```

#### Removing stopwords (common words)

```python
# Remove stopwords
import nltk  # Natural Language Toolkit
```

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
```

```python
len(stopwords.words('english'))  # Total stopwords in english
```

```python
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    x = []
    for wrd in text.split():
        if wrd not in stop_words:
            x.append(wrd)

    return x
```

```python
df['cleaned_review'] = df['cleaned_review'].apply(remove_stopwords)
```

```python
df.sample(4)
```

#### Stemming, or lemmatizing the words

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stem_words(text):
    x = []
    for wrd in text:
        x.append(ps.stem(wrd))

    return x
```

```python
df['cleaned_review'] = df['cleaned_review'].apply(stem_words)
```

```python
df.head()
```

#### Again joining words and create review like string

```python
def join_back(list_input):
    return " ".join(list_input)
```

```python
df['cleaned_review'] = df['cleaned_review'].apply(join_back)
```

#### Convert reviews from number to a dictionary of words

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=1200)
```

```python
x_vectorized = cv.fit_transform(x).toarray()
```

```python
y = df.iloc[:, 1].values
```

## Starts training the model

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_vectorized, y, test_size=0.2, random_state=18)
```

```python
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
```

```python
gaussian_NB = GaussianNB()
multi_NB = MultinomialNB()
bernoulli_NB = BernoulliNB()
```

```python
gaussian_NB.fit(x_train, y_train)
```

```python
multi_NB.fit(x_train, y_train)
```

```python
bernoulli_NB.fit(x_train, y_train)
```

```python
gaussian_pred = gaussian_NB.predict(x_test)
```

```python
multi_pred = multi_NB.predict(x_test)
```

```python
bernoulli_pred = bernoulli_NB.predict(x_test)
```

#### Testing our model

```python
from sklearn.metrics import accuracy_score, precision_score
```

```python
print("Accuracy -")
print(f"Gaussian :", accuracy_score(y_test, gaussian_pred))
print(f"Multinomial :", accuracy_score(y_test, multi_pred))
print(f"Bernoulli :", accuracy_score(y_test, bernoulli_pred))
```

```python
print("Precision -")
print(f"Gaussian :", precision_score(y_test, gaussian_pred))
print(f"Multinomial :", precision_score(y_test, multi_pred))
print(f"Bernoulli :", precision_score(y_test, bernoulli_pred))
```
