### Data Cleaning Process
#### 1. Sample 10000 rows
#### 2. Remove HTML Tags
#### 3. Remove special characters
#### 4. Converting every words into lower case
#### 5. Removing common words
#### 6. Stemming (watching -> watch)

```python
df = data.sample(10000).reset_index(drop=True)
```

```python
df['sentiment'].replace({'positive': 1, 'negative': 0}, inplace=True)
```

```python
df.sample(2)
```

```python
import re
clean = re.compile("<.*?>")
re.sub(clean, ' ', df.iloc[2].review)
```

```python
# Function to clean HTML Tags

def clean_html(text):
    clean = re.compile("<.*?>")
    return re.sub(clean, ' ', text)
```

```python
df['cleaned_review'] = df['review'].apply(clean_html)
```

```python
df.sample()
```

```python
# Function to convert to lower case

def convert_lower(text):
    return text.lower()
```

```python
df['cleaned_review'] = df['cleaned_review'].apply(convert_lower)
```

```python
# Function to remove special character

def remove_special(text):
    x = ''
    for ch in text:
        if ch.isalnum():
            x = x+ch
        else:
            x = x+' '
            
    return x
```

```python
df['cleaned_review'] = df['cleaned_review'].apply(remove_special)
```

```python
# Remove stopwords
import nltk  # Natural Language Toolkit
```

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
```

```python
len(stopwords.words('english'))  # Total stopwords in english
```

```python
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    x = []
    for wrd in text.split():
        if wrd not in stop_words:
            x.append(wrd)

    return x
```

```python
df['cleaned_review'] = df['cleaned_review'].apply(remove_stopwords)
```

```python
df
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stem_words(text):
    x = []
    for wrd in text:
        x.append(ps.stem(wrd))

    return x
```

```python
df['cleaned_review'] = df['cleaned_review'].apply(stem_words)
```

```python
df
```

```python
# Join back

def join_back(list_input):
    return "".join(list_input)
```

```python
df['cleaned_review'] = df['cleaned_review'].apply(join_back)
```
