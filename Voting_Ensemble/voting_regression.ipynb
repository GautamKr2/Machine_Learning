# Voting Regression

### Loading dataset

```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_california_housing
```

```python
x, y = fetch_california_housing(return_X_y=True)
```

```python
x.shape, y.shape
```

### Importing required model

```python
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import cross_val_score
```

```python
lin_reg = LinearRegression()
svr = SVR()
dt_reg = DecisionTreeRegressor()
```

### Finding individual r2-score

```python
estimators = [('lr', lin_reg), ('svr', svr), ('dtr', dt_reg)]
```

```python
for estimator in estimators:
    scores = cross_val_score(estimator[1], x, y, cv=10, scoring='r2')
    print(estimator[0], np.mean(scores))
```

### Using ensemble learning

```python
from sklearn.ensemble import VotingRegressor
```

```python
vot_reg = VotingRegressor(estimators=estimators)
```

```python
score = cross_val_score(vot_reg, x, y, cv=10, scoring='r2')
```

```python
print(f'R-2 Score for vot_reg = {np.mean(score)}')
```

### Weighted Voting

```python
idx = np.random.randint(0, len(x), 1000)
```

```python
x1 = x[idx]
y1 = y[idx]
```

```python
x1.shape, y1.shape
```

```python
for i in range(1, 4):
    for j in range(1, 4):
        for k in range(1, 4):
            vot_re = VotingRegressor(estimators=estimators, weights=[i,j,k])
            scores = cross_val_score(vot_reg, x1, y1, cv=10, scoring='r2')
            print('For lr={}, svm={}, dt={}'.format(i, j, k), np.mean(scores))
```
