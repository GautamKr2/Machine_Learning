# Classification metrics for **more than two classes**

### Loading data

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('E:/Dataset/Iris.csv')
```

```python
df.sample(3)
```

```python
x = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

### Encoding categorical data

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()
```

```python
encoder.fit(y)
```

```python
encoder.classes_
```

```python
y_encoded = encoder.transform(y)
```

```python
y_encoded
```

### Splitting data

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y_encoded, test_size=0.2, random_state=17)
```

### Training model

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

```python
log_reg = LogisticRegression()
dec_tree = DecisionTreeClassifier()
```

```python
log_reg.fit(x_train, y_train)
dec_tree.fit(x_train, y_train)
```

```python
y_pred1 = log_reg.predict(x_test)
y_pred2 = dec_tree.predict(x_test)
```

### Calculating metrices

```python
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
print(f"Accuracy(Logistic Regression): {accuracy_score(y_test, y_pred1)}")
print(f"Accuracy(Decision Tree): {accuracy_score(y_test, y_pred2)}")
```

```python
print(f"Confusion Matrics (Logistic Regression):\n {confusion_matrix(y_test, y_pred1)}")
print(f"\nConfusion Matrix(Decision Tree):\n {confusion_matrix(y_test, y_pred2)}")
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score
```

```python
print(f"Precision(Logistic Regression): {precision_score(y_test, y_pred1, average=None)}")
print(f"Precision(Decision Tree): {precision_score(y_test, y_pred2, average=None)}")

print(f"\nRecall(Logistic Regression): {recall_score(y_test, y_pred1, average=None)}")
print(f"Recall(Decision Tree): {recall_score(y_test, y_pred2, average=None)}")

print(f"\nF1 Score(Logistic Regression): {f1_score(y_test, y_pred1, average=None)}")
print(f"F1 Score(Decision Tree): {f1_score(y_test, y_pred2, average=None)}")
```

```python
from sklearn.metrics import classification_report
```

```python
print(f"Classification Report for Logistic Regression:\n\n {classification_report(y_test, y_pred1)}")
```

## On mnist dataset (digit classification)

```python
df1 = pd.read_csv('E:/Dataset/digit_recognition.csv')
```

```python
df1.sample(2)
```

```python
x1 = df1.iloc[:, 1:]
y1 = df1.iloc[:, 0]
```

```python
x_train1, x_test1, y_train1, y_test1 = train_test_split(x1, y1, test_size=0.2, random_state=17)
```

#### Model - 2

```python
log_reg1 = LogisticRegression()
dec_tree1 = DecisionTreeClassifier()
```

```python
log_reg1.fit(x_train1, y_train1)
dec_tree1.fit(x_train1, y_train1)
```

```python
y_pred3 = log_reg1.predict(x_test1)
y_pred4 = dec_tree1.predict(x_test1)
```

```python
print(f"Accuracy(Logistic Regression): {accuracy_score(y_test1, y_pred3)}")
print(f"Accuracy(Decision Tree): {accuracy_score(y_test1, y_pred4)}")
```

```python
print(f"Confusion Matrics (Logistic Regression):\n {confusion_matrix(y_test1, y_pred3)}")
print(f"\nConfusion Matrix(Decision Tree):\n {confusion_matrix(y_test1, y_pred4)}")
```

```python
print(f"Precision(Logistic Regression):\n {precision_score(y_test1, y_pred3, average=None)}")
print(f"\nPrecision(Decision Tree):\n {precision_score(y_test1, y_pred4, average=None)}")

print(f"\n\nRecall(Logistic Regression):\n {recall_score(y_test1, y_pred3, average=None)}")
print(f"\nRecall(Decision Tree):\n {recall_score(y_test1, y_pred4, average=None)}")

print(f"\n\nF1 Score(Logistic Regression):\n {f1_score(y_test1, y_pred3, average=None)}")
print(f"\nF1 Score(Decision Tree):\n {f1_score(y_test1, y_pred4, average=None)}")
```

```python
print("Classification Report (Logistic Regression):\n")
print(classification_report(y_test1, y_pred3))
```

```python
print("Classification Report (Decision Tree):\n")
print(classification_report(y_test1, y_pred4))
```
